```python
import torch

if torch.cuda.is_available():
    device=torch.device("cuda")
else:
    device=torch.device("cpu")

print(device)
```

```python
print(torch.cuda.get_device_name(0))
```

```python
my_tensor=torch.ones(5,3).to(device)
print(my_tensor)
```

```python
my_tensor2=torch.ones(5,3,requires_grad=True)
print(my_tensor2)
```

```python
x=torch.randn(5,3,requires_grad=True)
x
```

```python
y=x+x+x
y
```

```python
z=y**y
z
```

```python
v=z.mean()
v
```

```python
v.backward()
print(x.grad)
```
