```python
import torch 
import torch.nn as nn 
```

```python
x=torch.tensor([[1],[2],[3],[4]],dtype=torch.float32)
y=torch.tensor([[2],[4],[6],[8]],dtype=torch.float32)

samples,features=x.shape
input  =features
output =features

test=torch.tensor([5],dtype=torch.float32)

class linear_regression(nn.Module):
    def __init__(self,input,output):
        super(linear_regression,self).__init__()
        self.layer1=nn.Linear(input,output)

    def forward(self,x):
        return self.layer1(x)

model=linear_regression(input,output)

print(f"before training we will try to get output of input 5 :{model(test).item():.3f}")

lr=0.01
epochs=200

loss=nn.MSELoss()
optimizer=torch.optim.SGD(model.parameters(),lr=lr)

for epoch in range(epochs):

    predicition=model(x)

    the_loss=loss(y,predicition)

    the_loss.backward()

    optimizer.step()

    optimizer.zero_grad()

    w,b=model.parameters()
    if epoch % 10 ==0:
        print(f" epoch{epoch} , w={w[0][0].item():.3f} , b={b[0].item():.3f} , loss={the_loss.item():.3f}")

    

print(f"after training we will try to get output of input 5 :{model(test).item():.3f}")
```
