# reshape && view


## first reshape

```python
my_torch=my_torch.reshape(2,5)
my_torch
```

```python
# if we don't know the num of columns
my_torch2=torch.arange(10)
my_torch2
```

```python
my_torch2=my_torch2.reshape(2,-1)
my_torch2
```

## second view

```python
my_torch3=torch.arange(10)
my_torch3
```

```python
my_torch3=my_torch3.view(2,5)
my_torch3
```
