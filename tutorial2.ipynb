```python
import torch , numpy as np
```

#Tensor Math Operations

```python
tensor_1=torch.tensor([1,2,3,4])
tensor_2=torch.tensor([5,6,7,8])
```

##addition

### we have two wayes first one :

```python
tensor_1+tensor_2
```

### second one

```python
torch.add(tensor_1,tensor_2)
```

## subtraction

### first method

```python
tensor_1-tensor_2
```

### second method

```python
torch.sub(tensor_1,tensor_2)
```

## multiplication

### first method

```python
tensor_1 * tensor_2
```

### second method

```python
torch.mul(tensor_1,tensor_2)
```

## devision

### first method

```python
tensor_1 / tensor_2
```

### second method

```python
torch.div(tensor_1,tensor_2)
```

## remainder

### first method

```python
tensor_1 % tensor_2
```

### second method

```python
torch.remainder(tensor_1,tensor_2)
```

## power

```python
torch.pow(tensor_1,tensor_2)
```
