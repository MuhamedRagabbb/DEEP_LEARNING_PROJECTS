during creation of the DataSet

https://pytorch.org/docs/stable/torchvision/transforms.html

---------

RandomCrop, RandomHorizontalFlip, RandomRotation

On Tensors
----------

Use Lambda 

Custom

Write own class

composed = transforms.Compose([Rescale(256),

```python
import torch
import torchvision
import numpy as np
import math 
from torch.utils.data import DataLoader ,Dataset
```

```python
class WineData(Dataset):
    def __init__(self,transform=None):
        xy=np.loadtxt(
            "./wine.csv",
            dtype=np.float32,
            delimiter= ",",
            skiprows=1
        )
        self.num_samples=xy.shape[0]
        self.transform=transform
        self.x_data=xy[:,1:]
        self.y_data=xy[:,[0]]


    def __getitem__(self, index):
        sample=self.x_data[index],self.y_data[index]
        if self.transform:
            #! here we can do transform(samples) because we use __call__
            sample=self.transform(sample)
        return sample


    def __len__(self):
        return self.num_samples



class toTensor:
    #! when you create object ex. x = toTensor  x.__call__() = x()
    def __call__(self,sample):
        inputs,outputs=sample
        return torch.from_numpy(inputs) ,torch.from_numpy(outputs)


class MUl:
    def __init__(self,x):
        self.x=x

    def __call__(self,sample):
        inputs,outputs=sample
        return inputs*self.x,outputs


data=WineData(transform=toTensor())
features,target=data[0]
print(features,target)
print("*"*200)


composed=torchvision.transforms.Compose([toTensor(),MUl(2)])
data2=WineData(transform=composed)
features2,target2=data2[0]
print(features2,target2)
print("*"*200)

dataloader=DataLoader(dataset=data2,shuffle=True,batch_size=4)
iterable=iter(dataloader)
next_batch=next(iterable)
features3,target3=next_batch
print(features3,target3)
print("*"*200)
```
