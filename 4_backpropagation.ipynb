# backpropagation using linear regression equation 

```python
import torch 
# افترض ان ال BIAS = 0
x=torch.tensor(1.0)
w=torch.tensor(1.0,requires_grad=True)
y=torch.tensor(2.0)

# forward pass 
y_hat=x*w
loss=(y_hat-y)**2
print(loss)
```

```python
# backward pass
loss.backward()
print(w.grad)
```
